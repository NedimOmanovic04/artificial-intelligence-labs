## a)

```python
white["label"] = 0
red["label"] = 1
wines = pd.concat([white, red])
wines.head()
```

## c)

```python
X = wines.loc[:, wines.columns != "label"].to_numpy()
y = wines["label"].to_numpy()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

## d)

```python
model = models.Sequential()
model.add(layers.Dense(8, activation="relu", input_shape=(12,)))
model.add(layers.Dense(8, activation="relu"))
model.add(layers.Dense(1, activation="sigmoid"))
```

## e)

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
```

## f)

```python
history = model.fit(X_train, y_train, epochs=20, batch_size=16)

print("Final accuracy:", history.history["accuracy"][-1])
print("Final loss:", history.history["loss"][-1])
```


Model je istreniran kroz 20 epoha. Na kraju treniranja postignuta je tačnost od X i vrijednost funkcije gubitka od Y, što pokazuje da model uspješno uči razlikovati crna i bijela vina.

## g)

```python
scaler = StandardScaler().fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

model2 = models.Sequential()
model2.add(layers.Dense(8, activation="relu", input_shape=(12,)))
model2.add(layers.Dense(8, activation="relu"))
model2.add(layers.Dense(1, activation="sigmoid"))

model2.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

history_scaled = model2.fit(X_train_scaled, y_train, epochs=20, batch_size=16)

print("Accuracy bez skaliranja:", history.history["accuracy"][-1])
print("Accuracy sa skaliranjem:", history_scaled.history["accuracy"][-1])
```

Standardizacija može poboljšati performanse modela jer dovodi sve značajke na sličnu skalu. Rezultati se upoređuju i vidi se uticaj skaliranja.

## h)

```python
loss, acc = model2.evaluate(X_test_scaled, y_test)

print("Test accuracy:", acc)
print("Test loss:", loss)
```

Model je evaluiran nad testnim skupom podataka. Postignuta tačnost iznosi X, što pokazuje da model uspješno generalizuje i može razlikovati crna i bijela vina na novim podacima.
