##b) Napisati Python program koji za početnu vrijednost n računa n-ti Fibonacci broj. Obavezno koristiti for petlju;

```python
n = int(input("Unesi n: "))

a, b = 0, 1

for i in range(n):
    a, b = b, a + b

print("Fibonacci broj je:", a)
```

##c)  Napisati Python program koji za zadane vrijednosti a, b, i c računa vrijednosti rješenja kvadratne jednačine

```python
import math

a = float(input("Unesi a: "))
b = float(input("Unesi b: "))
c = float(input("Unesi c: "))

D = b**2 - 4*a*c

if D < 0:
    print("Rješenja su kompleksna!")
elif D == 0:
    x = -b / (2*a)
    print("Rješenja su ista!")
    print("x =", x)
else:
    x1 = (-b + math.sqrt(D)) / (2*a)
    x2 = (-b - math.sqrt(D)) / (2*a)
    print("x1 =", x1)
    print("x2 =", x2)
```

##d) Definisati funkciju fact(n) koja računa faktorijel broja n. Zatim definisati funkciju choose(n, k) koja
računa.Demonstrirati rad ovih funkcija na primjeru.

```python
def fact(n):
    rezultat = 1
    for i in range(1, n+1):
        rezultat *= i
    return rezultat

def choose(n, k):
    return fact(n) / (fact(k) * fact(n-k))

print("5! =", fact(5))
print("C(5,2) =", choose(5,2))
```

##g)Definisati funkciju kvadratna(M) koja prima NumPy 1D niz, te isti preoblikuje u kvadratnu matricu, ukoliko je to moguće. Ukoliko nije moguće, funkcija ne treba da uradi ništa.

```python
def kvadratna(niz):
    n = len(niz)
    korijen = int(np.sqrt(n))

    if korijen * korijen == n:
        return np.array(niz).reshape(korijen, korijen)
    else:
        return None


# primjer
niz = [1,2,3,4]
print(kvadratna(niz))

niz2 = [1,2,3,4,5]
print(kvadratna(niz2))
```

#Zadatak 4 - Rad sa grafikom
##a) Neka je data funkcija x = sin(n) na 100 tačaka (n = 1:100). Iscrtati linijski dijagram ove funkcije. Postaviti odgovarajuće labele, te naslov. Zatim preko tog dijagrama (na isti graf) nacrtati funkciju y = cos(n), također na 100 uzoraka. Uključiti mrežu (grid);

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(1, 101)
x = np.sin(n)
y = np.cos(n)

plt.plot(n, x, label='sin(n)')
plt.plot(n, y, label='cos(n)')

plt.xlabel('n')
plt.ylabel('Vrijednost')
plt.title('Grafik funkcija sin(n) i cos(n)')
plt.legend()
plt.grid(True)

plt.show()
```

##b) Ponoviti prethodni zadatak, ali sada iscrtati u istom prozoru (ali na različitim graficima) signale x i y. Postaviti odgovarajuće labele i naslove za oba grafika;

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(1, 101)
x = np.sin(n)
y = np.cos(n)

plt.figure()

plt.subplot(2,1,1)
plt.plot(n, x)
plt.title('sin(n)')
plt.xlabel('n')
plt.ylabel('sin(n)')
plt.grid(True)

plt.subplot(2,1,2)
plt.plot(n, y)
plt.title('cos(n)')
plt.xlabel('n')
plt.ylabel('cos(n)')
plt.grid(True)

plt.tight_layout()
plt.show()
```
